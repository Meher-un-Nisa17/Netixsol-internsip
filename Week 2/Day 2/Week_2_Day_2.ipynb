### LangChain Core Concepts

LangChain provides abstractions that automate several components of a raw-Python AI agent. The LLM wrapper (`ChatHuggingFace`) handles interaction with the language model, tools wrap Python functions so they can be used by an agent, `AgentExecutor` manages agent and tool execution, and memory/message-history components manage conversational context.

### LCEL Pipe Syntax

LCEL's `|` syntax connects runnable components into a sequence, where the output of one component becomes the input of the next. In this example, `prompt | llm` formats the user's question into a chat prompt and passes it directly to the Hugging Face language model when the chain is invoked.


# **TASK 2**

```python
from langchain_core.tools import tool
import pandas as pd

@tool
def calculator(expression: str) -> float:
    """Evaluate a basic arithmetic expression containing numbers and operators."""
    return eval(expression)


@tool
def get_weather(city: str) -> dict:
    """Return the temperature and weather condition for a city."""
    weather_data = {
        "Lahore": {"temperature": 35, "condition": "Sunny"},
        "Islamabad": {"temperature": 29, "condition": "Cloudy"},
        "Karachi": {"temperature": 32, "condition": "Sunny"}
    }

    if city in weather_data:
        return weather_data[city]

    return {"error": f"Weather data not available for {city}"}


@tool
def read_csv(file_path: str) -> str:
    """Read a CSV file from the local filesystem and return its contents as text."""
    df = pd.read_csv(file_path)
    return df.to_string(index=False)


tools = [calculator, get_weather, read_csv]

print("Tools created successfully.")
```

### create csv

```python
products = pd.DataFrame({
    "product": ["Laptop", "Headphones", "Keyboard", "Mouse"],
    "price": [120000, 8500, 4500, 2500],
    "category": ["Computer", "Audio", "Accessories", "Accessories"]
})

products.to_csv("products.csv", index=False)

print(products)
```

```python
print("Calculator:")
print(calculator.invoke("120000 * 0.90"))

print("\nWeather:")
print(get_weather.invoke("Lahore"))

print("\nCSV:")
print(read_csv.invoke("products.csv"))
```

```python
for tool_item in tools:
    print(f"\nTool: {tool_item.name}")
    print(f"Description: {tool_item.description}")
    print(f"Arguments: {tool_item.args}")
```

### Tool Docstrings

Tool docstrings are important in LangChain because they become part of the tool's metadata that is provided to the language model. The model uses the tool name, description, and input schema to understand what each tool does, decide when a tool is appropriate, and determine what arguments it should provide.

For example, the docstring for `read_csv` *"Read a CSV file from the local filesystem and return its contents as text"* tells the model that this tool is designed for reading CSV files and that it should be considered when the user asks about information stored in a CSV file.

Therefore, tool docstrings should be clear, specific, and descriptive because they influence the model's toolselection decisions.

## Task 3 — Agent Execution Trace Analysis

The LangChain agent was tested with a multi-step task:

> Read `products.csv`, find the price of the Laptop, then calculate the Laptop price after a 10% discount.

### Trace Annotation

The execution followed the following pattern:

1. **Reason**
   - The agent determined that it needed information from `products.csv` before performing the calculation.

2. **Act**
   - The agent selected the `read_csv` tool and supplied `products.csv` as the input.

3. **Observe**
   - The tool returned the contents of the CSV file, including the Laptop price of 120,000.

4. **Reason**
   - Using the observed Laptop price, the agent determined that it needed to calculate a 10% discount.

5. **Act**
   - The agent selected the `calculator` tool and supplied the appropriate arithmetic expression.

6. **Observe**
   - The calculator returned the discounted price of 108,000.

7. **Final Answer**
   - The agent used the tool results to produce the final response.

### Comparison with the Raw-Python Agent

The LangChain trace is similar to the raw-Python agent because both follow the same fundamental agent loop:

**Reason --> Act --> Observe --> Reason --> Act --> Observe --> Final Answer**

In the raw-Python implementation, the agent logic, tool calls, tool results, and message handling were explicitly managed in Python code. In LangChain, `AgentExecutor` manages this orchestration automatically, while `verbose=True` exposes the major execution steps.

### What is Hidden by LangChain?

LangChain hides several low-level implementation details from the developer. The framework manages the agent loop, tool-call routing, intermediate messages, and continuation of the conversation between tool calls. The model's private chain-of-thought/reasoning is not exposed; the verbose trace shows observable actions and tool results rather than the model's private reasoning.

Therefore, LangChain provides a higher-level abstraction that requires less orchestration code, while the raw-Python agent provides more direct control and visibility over the underlying API interaction.

### Conversation Memory

Conversation memory was added using `ConversationBufferMemory`. The memory stores previous messages and makes them available to the agent during subsequent turns.

The agent was tested with a three-turn conversation:

1. **"Find the price of the Laptop in products.csv."**
   - The agent used the `read_csv` tool and identified the Laptop price as 120,000.

2. **"Now compare it to the Headphones."**
   - The word "it" depended on the previous turn. The agent used the conversation history to understand that "it" referred to the Laptop and compared it with the Headphones priced at 8,500.

3. **"Which one should I recommend to a budget-conscious client?"**
   - The agent used the context from the previous turns to identify the two products being compared and recommended the lower-priced Headphones for a budget-conscious client.

### Memory vs. the Previous Agent

Without memory, each invocation would be treated as an independent request, so references such as "it", "the other one", or "which one" could be ambiguous.

With `ConversationBufferMemory`, previous messages are retained and supplied to the agent on subsequent turns. This allows the agent to maintain conversational context across multiple interactions.

The memory component therefore acts as the LangChain equivalent of the manually maintained conversation/message history used in the raw-Python agent.

Structured Output Model

```python
from pydantic import BaseModel, Field

class AgentResponse(BaseModel):
    answer: str = Field(description="The final answer to the user's request.")
    recommendation: str = Field(
        description="The recommended choice, or 'None' if no recommendation is needed."
    )
    reasoning_summary: str = Field(
        description="A brief summary of why the answer or recommendation was given."
    )

print("Structured output schema created.")
```

```python
structured_llm = llm.with_structured_output(AgentResponse)

structured_test = structured_llm.invoke(
    """The Laptop costs 120000 and the Headphones cost 8500.
    A budget-conscious client should choose the Headphones."""
)

print(structured_test)
```

Create new tool

```python
from langchain_core.tools import tool

@tool
def unreliable_product_lookup(product: str) -> str:
    """Look up a product price, but may fail because of a simulated data-source error."""

    try:
        product_prices = {
            "Laptop": 120000,
            "Headphones": 8500,
            "Keyboard": 4500,
            "Mouse": 2500
        }

        # Simulate a tool failure for one product
        if product == "Keyboard":
            raise RuntimeError("Product database temporarily unavailable.")

        if product not in product_prices:
            raise ValueError(f"Product '{product}' was not found.")

        return f"{product}: {product_prices[product]}"

    except Exception as e:
        return f"TOOL_ERROR: Unable to retrieve '{product}'. Reason: {str(e)}"
```

```python
tools_with_error = [
    calculator,
    get_weather,
    read_csv,
    unreliable_product_lookup
]

print([tool.name for tool in tools_with_error])
```

recreate the agent

```python
agent_with_errors = create_tool_calling_agent(
    llm=llm,
    tools=tools_with_error,
    prompt=agent_prompt
)
agent_executor_with_errors = AgentExecutor(
    agent=agent_with_errors,
    tools=tools_with_error,
    memory=memory,
    verbose=True
)

print("Error-handling agent with memory created successfully.")
```

```python
result = agent_executor_with_errors.invoke({
    "input": "Use the product lookup tool to find the price of the Laptop."
})

print("\nFinal Answer:")
print(result["output"])
```

```python
result = agent_executor_with_errors.invoke({
    "input": (
        "Use the product lookup tool to find the price of the Keyboard. "
        "If the lookup fails, explain what happened and give me the "
        "available product prices from products.csv instead."
    )
})

print("\nFinal Answer:")
print(result["output"])
```

Structured Output & Error Handling

### Structured Output

A Pydantic `AgentResponse` model was created to define a predictable structure for the agent's final response. LangChain's `with_structured_output()` converted the agent's normal response into the required schema containing `answer`, `recommendation`, and `reasoning_summary`. The resulting output was returned as a structured Python object rather than an unstructured text response.

### Error Handling

An `unreliable_product_lookup` tool was created to simulate a temporary product database failure. When the lookup for `Keyboard` failed, the tool caught the exception and returned a `TOOL_ERROR` message instead of terminating the agent. The agent recognized the failed lookup and recovered by using the `read_csv` tool to retrieve the Keyboard price from `products.csv`, demonstrating graceful fallback behavior.

### Reflection

LangChain made tool registration, agent orchestration, conversation memory, error recovery, and structured response handling much easier than the raw Python agent from Day 1. Instead of manually managing tool calls, observations, message history, and the agent loop, `AgentExecutor` coordinated these operations automatically. The main abstraction leak was that much of the orchestration happens behind the framework, so debugging required reading the verbose execution trace to understand which tools were called and why. I also noticed that graceful error recovery was not completely automatic; the tool needed explicit exception handling so that the agent could receive the failure as an observation and choose a fallback.
