# **TASK 1**

Multi-Agent Design Thinking

### Business Task

**Analyze a product dataset and produce a recommendation for a budget conscious client.**

The system will read product information from a CSV file, compare products based on price and category, and then produce a clear business recommendation.

### Agent 1: Product Research Specialist

**Role:** Product Research Specialist

**Goal:** Accurately extract and summarize product information from the provided CSV file, including product names, prices, and categories.

**Backstory:**
You are a careful product researcher who focuses on collecting accurate and factual information. You examine the available product data and organize it clearly so that other team members can use it for analysis.

**Responsibility:**
Read the product dataset and provide the factual product information needed by the analyst. This agent does not make recommendations or perform the final analysis.

---

### Agent 2: Product Pricing and Comparison Analyst

**Role:** Product Pricing and Comparison Analyst

**Goal:** Compare products using their prices and categories, perform relevant calculations, and identify the most suitable options for a budget-conscious client.

**Backstory:**
You are a detail-oriented business analyst specializing in pricing and product comparisons. You use numerical calculations and the researcher's findings to identify meaningful differences between products and support evidence-based decisions.

**Responsibility:**
Analyze the researcher's data, calculate price differences where useful, and determine which products provide the most suitable budget options. This agent does not write the final client-facing recommendation.

---

### Agent 3: Business Recommendation Writer

**Role:** Business Recommendation Writer

**Goal:** Transform the research and analysis into a clear, concise, and professional recommendation for the client.

**Backstory:**
You are a professional business writer who specializes in turning analytical findings into easy-to-understand recommendations. You rely only on the information provided by the research and analysis agents and avoid introducing unsupported claims.

**Responsibility:**
Combine the research and analysis into the final stakeholder-ready recommendation. This agent focuses on communication and presentation rather than performing new calculations or collecting data.

### Why Use Multiple Specialists?

Multiple specialists can outperform one generalist because each agent focuses on a specific stage of the task. The research agent ensures accurate data collection, the analyst focuses on numerical comparison, and the writer turns the findings into a clear recommendation. This separation can improve accuracy, organization, and consistency.

However, multiple agents are not always better. For a simple task with very little data or only one step, a single agent may be faster, cheaper, and easier to manage than coordinating several agents.


# **TASK 2**

```python
from crewai.tools import tool
import pandas as pd

@tool("read_csv")
def read_csv(file_path: str) -> str:
    """Reads a CSV file and returns its contents as text."""
    df = pd.read_csv(file_path)
    return df.to_string(index=False)


@tool("calculator")
def calculator(expression: str) -> str:
    """Evaluates a basic arithmetic expression and returns the numeric result."""
    try:
        result = eval(expression, {"__builtins__": {}}, {})
        return str(result)
    except Exception as e:
        return f"Calculation error: {e}"


print("Tools created successfully")
```

```python
from crewai import LLM

crew_llm = LLM(
    model="huggingface/openai/gpt-oss-120b",
    api_key=HF_TOKEN,
    base_url="https://router.huggingface.co/v1",
    temperature=0,
    prompt_cache=False
)

print("CrewAI LLM created successfully")
```

```python
from crewai import Agent

# Agent 1: Research
research_agent = Agent(
    role="Product Research Specialist",
    goal="Accurately extract and summarize product information from the provided CSV file.",
    backstory=(
        "You are a careful product researcher who focuses on factual information. "
        "You examine product data and provide clear findings about products, prices, and categories."
    ),
    tools=[read_csv],
    llm=crew_llm,
    verbose=True
)

# Agent 2: Analyst
analyst_agent = Agent(
    role="Product Pricing and Comparison Analyst",
    goal="Compare products using their prices and categories and identify the best options for a budget-conscious client.",
    backstory=(
        "You are a detail-oriented business analyst specializing in pricing comparisons. "
        "You use numerical calculations to identify price differences and meaningful comparisons."
    ),
    tools=[calculator],
    llm=crew_llm,
    verbose=True
)

# Agent 3: Writer
writer_agent = Agent(
    role="Business Recommendation Writer",
    goal="Turn research and analysis findings into a clear and practical product recommendation.",
    backstory=(
        "You are a professional business writer who creates concise, easy-to-understand recommendations. "
        "You rely on provided research and analysis rather than inventing information."
    ),
    tools=[],
    llm=crew_llm,
    verbose=True
)

print("3 agents created successfully")
```

### Build Agents & Assign Tools Justification

Three CrewAI agents were implemented with separate roles, goals, backstories, and LLM configuration.

| Agent                                  | Assigned Tool    | Reason                                                                                            |
| -------------------------------------- | ---------------- | ------------------------------------------------------------------------------------------------- |
| Product Research Specialist            | `read_csv`       | Needs direct access to `products.csv` to retrieve accurate product names, prices, and categories. |
| Product Pricing and Comparison Analyst | `calculator`     | Needs numerical calculations for price differences and comparisons.                               |
| Business Recommendation Writer         | No external tool | Only needs the research and analysis outputs to create the final recommendation.                  |

Tool access was intentionally kept role-specific instead of giving every agent all available tools. This reduces unnecessary tool usage and keeps responsibilities clearly separated. The Research Agent collects information, the Analyst performs numerical comparisons, and the Writer focuses on communicating the final recommendation.

All three agents use the CrewAI-compatible `crew_llm` configuration connected to the Hugging Face `gpt-oss-120b` model.


# **TASK 3**

### Building Tasks

```python
from crewai import Task

research_task = Task(
    description=(
        "Read the products.csv file and identify all available products, "
        "their prices, and categories. Provide accurate factual information "
        "that can be used by the analyst."
    ),
    expected_output=(
        "A clear list or table containing every product, its price, and category, "
        "with no invented information."
    ),
    agent=research_agent
)

analysis_task = Task(
    description=(
        "Using the research provided by the previous agent, compare the products "
        "based on price and category. Calculate useful price differences where "
        "appropriate and identify which products are most suitable for a "
        "budget-conscious client."
    ),
    expected_output=(
        "A concise comparison of the products, including relevant price "
        "calculations and a clear identification of the most budget-friendly options."
    ),
    agent=analyst_agent,
    context=[research_task]
)

writing_task = Task(
    description=(
        "Using the research and analysis provided by the previous agents, "
        "write a final recommendation for a budget-conscious client. "
        "Mention the relevant products, prices, categories, and reasoning. "
        "Do not invent information that is not present in the provided outputs."
    ),
    expected_output=(
        "A clear, professional final recommendation that summarizes the "
        "product comparison and explains which option or options are best "
        "for a budget-conscious client."
    ),
    agent=writer_agent,
    context=[research_task, analysis_task]
)

print("3 tasks created successfully")
```

### Assembling Crew

```python
from crewai import Crew, Process

crew = Crew(
    agents=[
        research_agent,
        analyst_agent,
        writer_agent
    ],
    tasks=[
        research_task,
        analysis_task,
        writing_task
    ],
    process=Process.sequential,
    verbose=True
)

print("Sequential crew created successfully")
```

### Running Crew

```python
import crewai.llms.cache as crewai_cache

def disable_cache_breakpoint(message):
    return message

crewai_cache.mark_cache_breakpoint = disable_cache_breakpoint

print("CrewAI cache_breakpoint disabled for this notebook")
```

```python
result = await crew.akickoff()

print("\n--- FINAL RESULT ---")
print(result)
```

During the review of the sequential execution, the Research Agent returned the product information as plain CSV-style text. Although the Analyst Agent could process it, the output was not as structured as desired for reliable downstream analysis. I improved the `expected_output` for the Research Task to explicitly require a clear list or table containing every product, price, and category, with no invented information. I also strengthened the Analyst Task instructions to perform only relevant price calculations and provide a concise comparison that the Writer could directly use. This made the handoff between agents clearer and reduced unnecessary or unsupported information in later outputs.


# **TASK 4**

### **Manger Agent**

```python
from crewai import Agent

manager_agent = Agent(
    role="Product Analysis Manager",
    goal=(
        "Coordinate the product research, pricing analysis, and recommendation "
        "process to produce an accurate and useful recommendation for a "
        "budget-conscious client."
    ),
    backstory=(
        "You are an experienced project manager who coordinates specialized "
        "agents. You review their work, ensure that information flows correctly "
        "between tasks, resolve inconsistencies, and make sure the final result "
        "is accurate, complete, and based only on the available product data."
    ),
    llm=crew_llm,
    verbose=True
)

print("Manager agent created successfully")
```

```python
from crewai import Task

hier_research_task = Task(
    description=(
        "Read the products.csv file using the available read_csv tool. "
        "Identify every product, its price, and category. "
        "Return accurate product data in a clear table or list. "
        "Do not invent or infer any information."
    ),
    expected_output=(
        "A clear table containing all products, their prices, and categories "
        "exactly as found in products.csv."
    )
)

hier_analysis_task = Task(
    description=(
        "Using the product research provided by the research specialist, "
        "compare the products by price and category. "
        "Use the calculator tool when numerical calculations are needed. "
        "Identify the most budget-friendly options."
    ),
    expected_output=(
        "A concise product comparison containing relevant price differences "
        "and a clear identification of the most budget-friendly option."
    ),
    context=[hier_research_task]
)

hier_writing_task = Task(
    description=(
        "Using the research and analysis from the other specialists, "
        "write a final recommendation for a budget-conscious client. "
        "Use only information supported by the previous outputs."
    ),
    expected_output=(
        "A clear, professional recommendation that includes the relevant "
        "products, prices, categories, and reasoning."
    ),
    context=[hier_research_task, hier_analysis_task]
)

print("Hierarchical tasks created successfully")
```

### Hierarchical Crew

```python
hierarchical_crew = Crew(
    agents=[
        research_agent,
        analyst_agent,
        writer_agent
    ],
    tasks=[
        hier_research_task,
        hier_analysis_task,
        hier_writing_task
    ],
    process=Process.hierarchical,
    manager_agent=manager_agent,
    verbose=True
)

print("Corrected hierarchical crew created successfully")
```

```python
import importlib
import litellm

# Reset LiteLLM to its original state
litellm = importlib.reload(litellm)

# Save the original completion function
_original_completion = litellm.completion

# Remove the unsupported parameter before sending the request
def completion_without_prompt_cache(*args, **kwargs):
    kwargs.pop("prompt_cache", None)
    return _original_completion(*args, **kwargs)

litellm.completion = completion_without_prompt_cache

print("LiteLLM reset and prompt_cache patch applied successfully")
```

```python
hierarchical_result = await hierarchical_crew.akickoff()

print("\n--- HIERARCHICAL FINAL RESULT ---")
print(hierarchical_result)
```

| Aspect             | Sequential                                                                                                                            | Hierarchical                                                                                                                                                  |
| ------------------ | ------------------------------------------------------------------------------------------------------------------------------------- | ------------------------------------------------------------------------------------------------------------------------------------------------------------- |
| **Quality**        | More consistent. Research → analysis → writing followed the intended roles, although the writer added a few unsupported descriptions. | More variable. It successfully delegated research, but the manager introduced a factual calculation error and unsupported claims in the final recommendation. |
| **Latency / cost** | Lower expected overhead because agents execute directly in a fixed pipeline.                                                          | Higher overhead because the manager must reason, delegate, receive responses, and coordinate workers.                                                         |
| **Reliability**    | Higher for this task. The workflow is predictable and dependencies are explicitly defined.                                            | Lower for this simple task. Delegation introduced extra steps and the manager initially could not access the CSV directly.                                    |
| **Pros**           | Simple, predictable, efficient, clear data flow, easy to debug.                                                                       | Flexible, dynamic delegation, manager can decide which specialist to involve and can coordinate complex work.                                                 |
| **Cons**           | Less flexible; task order is fixed and does not dynamically adapt.                                                                    | More complex, more LLM calls, greater token usage/latency, and more opportunities for delegation or reasoning errors.                                         |
| **When to use**    | Best for well-defined tasks with a known sequence of steps and clear dependencies.                                                    | Best for complex projects where tasks are less predictable and a manager needs to dynamically coordinate multiple specialists.                                |

###Evaluation & Cost Awareness

Three success criteria were used to evaluate the CrewAI output: factual grounding, completeness, and clarity and tone. Factual grounding checks whether prices, categories, calculations, and recommendations are supported by the product data; completeness checks whether the relevant products and recommendation are covered; and clarity and tone checks whether the response is understandable and professionally written.

The first sequential run used 348 total tokens across 3 successful requests. The hierarchical run used 51,723 total tokens across 37 successful requests, while the third sequential run used 3,105 tokens across 9 successful requests and took 8.97 seconds. Monetary cost was not directly calculated because the model was accessed through the Hugging Face router and the applicable provider pricing was not fixed for this experiment, so token usage was used as the cost-awareness measure.

For this specific product-recommendation task, the multi-agent crew was not worth the added complexity and cost compared with a single well-designed agent or a simple sequential workflow. The hierarchical crew used substantially more tokens and requests without producing a clear improvement in quality. It also introduced an incorrect price comparison and unsupported claims in its final response. For simple, predictable tasks like this one, sequential execution or a single well-designed agent is more efficient and reliable, while hierarchical multi-agent systems are more appropriate for complex tasks requiring dynamic delegation.
