### Retrieval Strategy

This AFL assistant uses structured retrieval as the primary retrieval method because the
available AFL data is structured into match-level, player-round, and player-season tables.

Structured Pandas lookups are used for:
- Team records and head-to-head results
- Player season statistics
- Match results and match-level statistics
- Exact numerical values such as wins, losses, disposals, goals, and fantasy points

This approach is preferred for sports statistics because exact values should be retrieved
directly from the dataset rather than generated through semantic similarity. Using fuzzy
semantic retrieval for numerical statistics could return related but incorrect records.

Semantic retrieval would be appropriate for unstructured AFL content such as match reports,
articles, commentary, or textual historical descriptions. No separate unstructured AFL text
dataset is currently being used, so a vector store is not required for this implementation.

Therefore, the retrieval layer is based on exact structured Pandas queries over the AFL
datasets.

### Building Tool 1 Team Record vs opponent

```python
# Canonical AFL team-name mapping for user-friendly lookups

TEAM_NAME_MAP = {
    "adelaide": "Adelaide Crows",
    "adelaide crows": "Adelaide Crows",

    "brisbane bears": "Brisbane Bears",

    "brisbane": "Brisbane Lions",
    "brisbane lions": "Brisbane Lions",

    "carlton": "Carlton Blues",
    "carlton blues": "Carlton Blues",

    "collingwood": "Collingwood Magpies",
    "collingwood magpies": "Collingwood Magpies",

    "essendon": "Essendon Bombers",
    "essendon bombers": "Essendon Bombers",

    "fitzroy": "Fitzroy Lions",
    "fitzroy lions": "Fitzroy Lions",

    "fremantle": "Fremantle Dockers",
    "fremantle dockers": "Fremantle Dockers",

    "geelong": "Geelong Cats",
    "geelong cats": "Geelong Cats",

    "gold coast": "Gold Coast Suns",
    "gold coast suns": "Gold Coast Suns",

    "gws": "Greater Western Sydney Giants",
    "greater western sydney": "Greater Western Sydney Giants",
    "greater western sydney giants": "Greater Western Sydney Giants",

    "hawthorn": "Hawthorn Hawks",
    "hawthorn hawks": "Hawthorn Hawks",

    "melbourne": "Melbourne Demons",
    "melbourne demons": "Melbourne Demons",

    "north melbourne": "North Melbourne Kangaroos",
    "north melbourne kangaroos": "North Melbourne Kangaroos",

    "port adelaide": "Port Adelaide Power",
    "port adelaide power": "Port Adelaide Power",

    "richmond": "Richmond Tigers",
    "richmond tigers": "Richmond Tigers",

    "st kilda": "St Kilda Saints",
    "st kilda saints": "St Kilda Saints",

    "sydney": "Sydney Swans",
    "sydney swans": "Sydney Swans",

    "west coast": "West Coast Eagles",
    "west coast eagles": "West Coast Eagles",

    "western bulldogs": "Western Bulldogs",
    "w. bulldogs": "Western Bulldogs",
}


def normalize_team_name(team):
    """
    Convert a user-provided AFL team name into the canonical
    team name used in the dataset.
    """
    team_clean = str(team).strip().lower()

    if team_clean in TEAM_NAME_MAP:
        return TEAM_NAME_MAP[team_clean]

    # Allow exact canonical names even if not explicitly mapped
    canonical_names = set(
        match_features["home_team"].dropna().unique()
    ) | set(
        match_features["away_team"].dropna().unique()
    )

    for canonical in canonical_names:
        if team_clean == str(canonical).lower():
            return canonical

    return None


def get_team_record_vs_opponent(team, opponent):
    """
    Retrieve the exact head-to-head record between two AFL teams
    from the reconstructed match-level dataset.

    The function searches both home and away matches and calculates
    wins, losses, and draws directly from the dataset.
    """

    team_original = str(team).strip()
    opponent_original = str(opponent).strip()

    team = normalize_team_name(team_original)
    opponent = normalize_team_name(opponent_original)

    # Validate team names
    if team is None:
        return {
            "error": f"Unknown AFL team: {team_original}"
        }

    if opponent is None:
        return {
            "error": f"Unknown AFL team: {opponent_original}"
        }

    if team == opponent:
        return {
            "error": "The team and opponent must be different."
        }

    # Team playing at home
    home_matches = match_features[
        (match_features["home_team"] == team) &
        (match_features["away_team"] == opponent)
    ]

    # Team playing away
    away_matches = match_features[
        (match_features["home_team"] == opponent) &
        (match_features["away_team"] == team)
    ]

    total_matches = len(home_matches) + len(away_matches)

    if total_matches == 0:
        return {
            "team": team,
            "opponent": opponent,
            "matches": 0,
            "wins": 0,
            "losses": 0,
            "draws": 0,
            "message": "No matching head-to-head records found in the dataset."
        }

    # Results when requested team was home
    home_wins = (
        home_matches["match_winner"] == "Home Win"
    ).sum()

    home_draws = (
        home_matches["match_winner"] == "Draw"
    ).sum()

    # Results when requested team was away
    away_wins = (
        away_matches["match_winner"] == "Away Win"
    ).sum()

    away_draws = (
        away_matches["match_winner"] == "Draw"
    ).sum()

    wins = int(home_wins + away_wins)
    draws = int(home_draws + away_draws)
    losses = int(total_matches - wins - draws)

    return {
        "team": team,
        "opponent": opponent,
        "matches": int(total_matches),
        "wins": wins,
        "losses": losses,
        "draws": draws
    }


# Test with short team names
result = get_team_record_vs_opponent(
    "Geelong",
    "Essendon"
)

print(result)
```

```python
# Create a player-name lookup from player_info

player_lookup = player_info[
    ["id", "player_name", "player_full_name"]
].copy()

player_lookup["player_name_lower"] = (
    player_lookup["player_name"]
    .fillna("")
    .astype(str)
    .str.strip()
    .str.lower()
)

player_lookup["player_full_name_lower"] = (
    player_lookup["player_full_name"]
    .fillna("")
    .astype(str)
    .str.strip()
    .str.lower()
)

print("Player lookup created.")
print("Number of players:", len(player_lookup))

```

```python
def get_player_season_stats(player_name, year=None):
    """
    Retrieve exact AFL player season statistics from the dataset.

    Parameters
    ----------
    player_name : str
        Player's common or full name.
    year : int, optional
        AFL season year. If omitted, all available seasons for the
        player are returned.

    Returns
    -------
    dict
        Exact player season statistics retrieved from the dataset.
    """

    query_name = str(player_name).strip().lower()

    # Search by common name or full name
    matches = player_lookup[
        (player_lookup["player_name_lower"] == query_name) |
        (player_lookup["player_full_name_lower"] == query_name)
    ]

    if matches.empty:
        return {
            "error": f"Player '{player_name}' was not found in the AFL dataset."
        }

    # Get matching player IDs
    player_ids = matches["id"].tolist()

    # Retrieve the player's seasonal records
    stats = player_seasonal_stats[
        player_seasonal_stats["player_id"].isin(player_ids)
    ].copy()

    # Optional season filtering
    if year is not None:
        try:
            year = int(year)
        except ValueError:
            return {
                "error": "Year must be a valid integer."
            }

        stats = stats[stats["year"] == year]

    if stats.empty:
        if year is not None:
            return {
                "player": player_name,
                "year": year,
                "message": "No season statistics found for this player and year."
            }

        return {
            "player": player_name,
            "message": "No season statistics found for this player."
        }

    # Columns exposed to the agent
    stat_columns = [
        "player_id",
        "year",
        "team",
        "is_finals",
        "games_played",
        "kicks",
        "marks",
        "handballs",
        "disposals",
        "goals",
        "behinds",
        "hit_outs",
        "tackles",
        "rebound_50s",
        "inside_50s",
        "clearances",
        "clangers",
        "free_kicks_for",
        "free_kicks_against",
        "brownlow_votes",
        "contested_possessions",
        "uncontested_possessions",
        "contested_marks",
        "marks_inside_50",
        "one_percenters",
        "bounces",
        "goal_assists",
        "total_score",
        "total_fantasy_points",
        "total_percentage_played",
        "avg_disposals",
        "avg_goals",
        "avg_tackles",
        "avg_fantasy_points"
    ]

    # Keep only columns that actually exist
    stat_columns = [
        col for col in stat_columns
        if col in stats.columns
    ]

    records = stats[stat_columns].to_dict(orient="records")

    return {
        "player": player_name,
        "records": records
    }
```

### Register AFL Retrival Tool

```python
from langchain_core.tools import tool


@tool
def team_record_tool(team: str, opponent: str) -> dict:
    """
    Retrieve the exact historical AFL record of one team against another team.

    Use this tool for questions about matches played, wins, losses, or draws
    between two AFL teams. The result comes directly from the AFL match dataset.
    """
    return get_team_record_vs_opponent(team, opponent)


@tool
def player_season_stats_tool(player_name: str, year: int = None) -> dict:
    """
    Retrieve exact AFL season statistics for a player.

    Use this tool for questions about a player's season statistics such as
    disposals, goals, tackles, games played, marks, or fantasy points.
    The result comes directly from the AFL player seasonal statistics dataset.
    """
    return get_player_season_stats(player_name, year)
```

### Scop Scoring

```python
# Corrected scope evaluation

def score_scope(category, response):
    response_lower = response.lower()

    refusal_indicators = [
        "i can only provide information related to the afl",
        "i can only provide information about the afl",
        "i'm sorry, but i can only provide",
        "i'm unable to provide information",
        "my focus is solely on afl",
        "i'm focused on afl",
        "i'm focused on afl-related",
        "i can only provide information related to afl players",
        "i can only provide information about afl teams",
        "i can only provide information related to australian football league"
    ]

    if category == "Legitimate AFL":
        # Legitimate AFL questions should receive an answer,
        # not an out-of-scope refusal.
        if any(indicator in response_lower for indicator in refusal_indicators):
            return "FAIL"
        return "PASS"

    else:
        # Off-topic and ambiguous questions should be refused,
        # redirected, or restricted to AFL.
        if any(indicator in response_lower for indicator in refusal_indicators):
            return "PASS"

        return "FAIL"


evaluation_df["scope_score"] = evaluation_df.apply(
    lambda row: score_scope(
        row["category"],
        row["response"]
    ),
    axis=1
)

evaluation_df[
    ["id", "category", "prompt", "scope_score"]
]
```

### Grounding Evaluation

```python
expected_grounding = {
    1: ["63", "39", "23", "1"],
    2: ["11"],
    3: ["1"],
    4: ["1"],
    5: ["1"],
    6: ["74"],
    7: ["4"],
    8: ["63"]
}


def score_grounding(test_id, response, expected_scope):
    if expected_scope != "AFL":
        return "NA"

    expected_values = expected_grounding.get(test_id, [])

    response_text = str(response)

    if all(value in response_text for value in expected_values):
        return "PASS"

    return "FAIL"


evaluation_df["grounding_score"] = evaluation_df.apply(
    lambda row: score_grounding(
        row["id"],
        row["response"],
        row["expected_scope"]
    ),
    axis=1
)

evaluation_df[
    ["id", "prompt", "scope_score", "grounding_score"]
]
```

### Evaluation Summary
